```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv(
    "../data/processed/iex_dam_2025_clean.csv",
    parse_dates=["timestamp"]
)

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Dataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(10, 5))

plt.hist(
    df["price_inr_mwh"],
    bins=50,
    edgecolor="black"
)

plt.title("Distribution of IEX Day-Ahead Market Prices - 2025")
plt.xlabel("Market Clearing Price (₹/MWh)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4))

plt.boxplot(
    df["price_inr_mwh"],
    vert=False
)

plt.title("MCP Price Distribution and Outliers")
plt.xlabel("Market Clearing Price (₹/MWh)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 5))

plt.plot(
    df["timestamp"],
    df["price_inr_mwh"],
    linewidth=0.5
)

plt.title("IEX DAM Market Clearing Price - 2025")
plt.xlabel("Date")
plt.ylabel("MCP (₹/MWh)")

plt.tight_layout()
plt.show()
```

```python
daily_prices = (
    df.groupby(df["timestamp"].dt.date)["price_inr_mwh"]
    .mean()
)

plt.figure(figsize=(15, 5))

plt.plot(daily_prices.index, daily_prices.values)

plt.title("Daily Average IEX DAM Price - 2025")
plt.xlabel("Date")
plt.ylabel("Average MCP (₹/MWh)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
hourly_prices = (
    df.groupby("hour")["price_inr_mwh"]
    .mean()
)

plt.figure(figsize=(11, 5))

plt.plot(
    hourly_prices.index,
    hourly_prices.values,
    marker="o"
)

plt.title("Average Electricity Price by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average MCP (₹/MWh)")

plt.xticks(range(24))
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(hourly_prices.sort_values(ascending=False))
```

```python
df["day_type"] = np.where(
    df["is_weekend"] == 1,
    "Weekend",
    "Weekday"
)

day_type_prices = (
    df.groupby("day_type")["price_inr_mwh"]
    .mean()
)

plt.figure(figsize=(7, 5))

plt.bar(
    day_type_prices.index,
    day_type_prices.values
)

plt.title("Average MCP: Weekday vs Weekend")
plt.ylabel("Average MCP (₹/MWh)")

plt.tight_layout()
plt.show()

print(day_type_prices)
```

```python
heatmap_data = df.pivot_table(
    values="price_inr_mwh",
    index="day_of_week",
    columns="hour",
    aggfunc="mean"
)

# Monday = 0, Sunday = 6
heatmap_data.index = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

plt.figure(figsize=(15, 6))

plt.imshow(
    heatmap_data,
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(label="Average MCP (₹/MWh)")

plt.title("Average Electricity Price: Day of Week vs Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Day of Week")

plt.xticks(range(24))
plt.yticks(range(7), heatmap_data.index)

plt.tight_layout()
plt.show()
```

```python
monthly_prices = (
    df.groupby("month")["price_inr_mwh"]
    .mean()
)

plt.figure(figsize=(11, 5))

plt.bar(
    monthly_prices.index,
    monthly_prices.values
)

plt.title("Monthly Average IEX DAM Price - 2025")
plt.xlabel("Month")
plt.ylabel("Average MCP (₹/MWh)")

plt.xticks(range(1, 13))

plt.tight_layout()
plt.show()

print(monthly_prices)
```

```python
price = df["price_inr_mwh"]

print("Maximum price:", price.max())

print("Records at ₹10,000/MWh:", (price == 10000).sum())

print(
    "Percentage at ₹10,000/MWh:",
    (price == 10000).mean() * 100
)

# Identify high-price observations
high_prices = df[
    df["price_inr_mwh"] >= 8000
]

print("\nObservations with MCP >= ₹8,000/MWh:")
display(high_prices.head(20))

print("\nTotal high-price observations:", len(high_prices))
```
